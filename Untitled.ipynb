```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the fps_log2.xlsx data (the one to be compressed)
fps_log2_data = pd.read_excel('fps_log2.xlsx')

# Clean the column names by stripping leading/trailing spaces
fps_log2_data.columns = fps_log2_data.columns.str.strip()

# Extract FPS data from the fps_log2.xlsx file
fps_log2_fps = fps_log2_data['FPS'].values

# Resample 2000 to 500 by averaging every 4 frames (skipping the first 4 zeros)
# Skipping the first 4 zeros, start from the 5th frame
fps_log2_compressed = np.array([np.mean(fps_log2_fps[i:i+4]) for i in range(4, len(fps_log2_fps)-4, 4)])

first_4_frames = fps_log2_fps[:4]  # These are the first 4 frames (usually non-zero)

# Concatenate the first 4 frames with the compressed data
compressed_fps_log2 = np.concatenate([first_4_frames, fps_log2_compressed[:500]])

# Load the fps_log1.xlsx data (comparison dataset)
fps_log1_data = pd.read_excel('fps_log1.xlsx')

# Clean the column names by stripping leading/trailing spaces
fps_log1_data.columns = fps_log1_data.columns.str.strip()

# Extract FPS data from the fps_log1.xlsx file
fps_log1_fps = fps_log1_data['FPS'].values

# Create the plot
plt.figure(figsize=(10, 6))

# Plot the compressed 500 FPS data from fps_log2 (in orange)
plt.plot(compressed_fps_log2, color='orange', label="Batch", linewidth=2)

# Plot the FPS data from fps_log1.xlsx (in green)
plt.plot(fps_log1_fps, label="Basic", linewidth=2)

# Add labels and title
plt.title("FPS Comparison", fontsize=16)
plt.xlabel("Frame Count in Real time Rendering", fontsize=14)
plt.ylabel("FPS", fontsize=14)

# Add legend
plt.legend(loc="upper right")

# Display grid
plt.grid(True)

# Show the plot
plt.show()
```
